# Form HTML e CSS: centratura e struttura di base

Questa guida mostra come centrare un form in orizzontale e in verticale nella pagina e come creare un form semplice, ordinato e adattabile agli schermi piccoli.

## 1. Il codice di partenza

Il form iniziale contiene tre campi e un pulsante di invio.

### index.html

```html
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
  <link rel="stylesheet" href="./index.css">
</head>
<body>

  <form action="">
    Name: <input type="text">
    Age: <input type="number">
    Address: <input type="text">
    <input type="submit">
  </form>

</body>
</html>
```

### index.css

```css
form {
  width: 200px;
  background-color: yellow;
  padding: 20px;
  border: 1px solid black;
  border-radius: 5px;
}

input {
  margin-bottom: 5px;
}
```

## 2. Centrare il form nella pagina

Per centrare il form in orizzontale e in verticale, possiamo usare **Flexbox sul body**, cioè sul contenitore del form.

Aggiungiamo questo blocco all'inizio di `index.css`, prima della regola `form`:

```css
body {
  margin: 0;
  min-height: 100vh;

  display: flex;
  justify-content: center;
  align-items: center;
}
```

### Il CSS completo con la modifica minima

```css
body {
  margin: 0;
  min-height: 100vh;

  display: flex;
  justify-content: center;
  align-items: center;
}

form {
  width: 200px;
  background-color: yellow;
  padding: 20px;
  border: 1px solid black;
  border-radius: 5px;
}

input {
  margin-bottom: 5px;
}
```

### Spiegazione delle proprietà

| Proprietà | Significato |
|---|---|
| `margin: 0` | Elimina il margine che il browser mette normalmente intorno al `body`. |
| `min-height: 100vh` | Fa occupare al `body` almeno tutta l'altezza della finestra. |
| `display: flex` | Attiva Flexbox per disporre gli elementi contenuti nel `body`. |
| `justify-content: center` | Con la direzione predefinita `row`, centra il form in orizzontale. |
| `align-items: center` | Con la direzione predefinita `row`, centra il form in verticale. |

### Che cosa significa 100vh?

`vh` è un'unità relativa all'altezza della viewport, cioè dell'area della finestra del browser che mostra la pagina.

- `1vh` corrisponde all'1% dell'altezza della viewport.
- `50vh` corrisponde alla metà dell'altezza della viewport.
- `100vh` corrisponde all'intera altezza della viewport.

Per esempio, se la viewport è alta 900 pixel:

```css
min-height: 100vh;
```

corrisponde a un'altezza minima di 900 pixel.

### Perché serve min-height?

Senza questa proprietà, il `body` normalmente è alto quanto il suo contenuto: non avrebbe spazio verticale libero nel quale centrare il form.

Con:

```css
min-height: 100vh;
```

il `body` occupa almeno tutta l'altezza della finestra. Flexbox può quindi posizionare il form al centro dello spazio disponibile.

Usiamo `min-height` invece di `height` perché, se il contenuto diventa più alto della finestra, la pagina può crescere.

### Attenzione alla direzione di Flexbox

Nel nostro esempio non abbiamo scritto `flex-direction`, quindi viene usato il valore predefinito:

```css
flex-direction: row;
```

Con questa direzione:

- `justify-content` agisce sull'asse orizzontale.
- `align-items` agisce sull'asse verticale.

Se impostiamo:

```css
flex-direction: column;
```

i due assi si scambiano:

- `justify-content` agisce sull'asse verticale.
- `align-items` agisce sull'asse orizzontale.

Quindi `justify-content` non significa sempre "allineamento orizzontale": dipende dalla direzione impostata.

## 3. Un form semplice e più ordinato

Possiamo migliorare il form aggiungendo:

- Un titolo.
- Etichette collegate ai campi.
- Un campo email.
- Indicazioni sui campi obbligatori.
- Spaziatura regolare.
- Un pulsante di invio.
- Una larghezza adattabile agli schermi piccoli.
- Un'indicazione visiva del campo selezionato.

Per provare i controlli di base non serve JavaScript.

Sostituiamo i due file iniziali con quelli seguenti.

## 4. File index.html completo

```html
<!DOCTYPE html>
<html lang="it">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Modulo di registrazione</title>
  <link rel="stylesheet" href="./index.css">
</head>

<body>

  <form method="get">
    <h1>Registrazione</h1>

    <p>I campi con * sono obbligatori.</p>

    <div class="campo">
      <label for="nome">Nome *</label>
      <input
        type="text"
        id="nome"
        name="nome"
        autocomplete="given-name"
        required
      >
    </div>

    <div class="campo">
      <label for="email">Email *</label>
      <input
        type="email"
        id="email"
        name="email"
        autocomplete="email"
        required
      >
    </div>

    <div class="campo">
      <label for="eta">Età</label>
      <input
        type="number"
        id="eta"
        name="eta"
        min="0"
        max="120"
      >
    </div>

    <div class="campo">
      <label for="indirizzo">Indirizzo</label>
      <input
        type="text"
        id="indirizzo"
        name="indirizzo"
        autocomplete="street-address"
      >
    </div>

    <button type="submit">Invia</button>
  </form>

</body>
</html>
```

## 5. File index.css completo

```css
* {
  box-sizing: border-box;
}

body {
  margin: 0;
  min-height: 100vh;
  padding: 20px;

  display: flex;
  justify-content: center;
  align-items: center;

  font-family: Arial, sans-serif;
  background-color: #f0f2f5;
  color: #222;
}

form {
  width: 100%;
  max-width: 360px;
  padding: 24px;

  background-color: white;
  border: 1px solid #ccc;
  border-radius: 8px;
}

h1 {
  margin: 0 0 10px;
  font-size: 26px;
}

p {
  margin: 0 0 20px;
  font-size: 14px;
  color: #555;
}

.campo {
  margin-bottom: 16px;
}

label {
  display: block;
  margin-bottom: 6px;
  font-weight: bold;
}

input {
  width: 100%;
  padding: 10px;

  border: 1px solid #999;
  border-radius: 5px;

  font: inherit;
}

input:focus {
  outline: 2px solid #155eab;
  outline-offset: 2px;
}

button {
  width: 100%;
  padding: 12px;

  background-color: #155eab;
  color: white;

  border: none;
  border-radius: 5px;

  font: inherit;
  cursor: pointer;
}

button:hover {
  background-color: #104a87;
}

button:focus-visible {
  outline: 2px solid #155eab;
  outline-offset: 3px;
}
```

## 6. Capire la struttura HTML

### Il contenitore di ogni campo

Ogni etichetta e il relativo input sono racchiusi in un `div`:

```html
<div class="campo">
  <label for="nome">Nome *</label>
  <input
    type="text"
    id="nome"
    name="nome"
    autocomplete="given-name"
    required
  >
</div>
```

La classe `campo` permette di gestire la distanza tra un gruppo e il successivo:

```css
.campo {
  margin-bottom: 16px;
}
```

### Il collegamento tra label e input

Consideriamo questo esempio:

```html
<label for="nome">Nome *</label>
<input type="text" id="nome" name="nome" required>
```

L'attributo:

```html
for="nome"
```

collega l'etichetta all'input che ha:

```html
id="nome"
```

I due valori devono coincidere.

Se clicchi sulla scritta "Nome", il cursore entra nel campo collegato. Questa associazione aiuta anche chi utilizza tecnologie assistive a capire a cosa serve il campo.

### La differenza tra id e name

Anche se nell'esempio hanno lo stesso valore, `id` e `name` hanno compiti diversi.

| Attributo | Funzione |
|---|---|
| `id="nome"` | Identifica l'elemento nella pagina e permette di collegarlo alla sua etichetta. Deve essere univoco nella pagina. |
| `name="nome"` | Indica la chiave con cui il valore del campo viene inviato dal form. |

Se scrivi Alessandro nel campo:

```html
<input type="text" id="nome" name="nome">
```

il dato inviato sarà:

```text
nome=Alessandro
```

Un campo senza `name` normalmente non viene incluso nei dati inviati dal form.

### I campi obbligatori

L'attributo `required` rende obbligatoria la compilazione:

```html
<input type="text" id="nome" name="nome" required>
```

Se provi a inviare il form lasciando il campo vuoto, il browser blocca l'invio e mostra un messaggio.

L'asterisco nell'etichetta:

```html
<label for="nome">Nome *</label>
```

è solo un'indicazione visiva. Il controllo vero è attivato da `required`.

### Il campo email

```html
<input
  type="email"
  id="email"
  name="email"
  autocomplete="email"
  required
>
```

`type="email"` attiva un controllo di base sul formato del valore inserito.

Per esempio:

```text
alessandro
```

non è un formato email valido.

```text
alessandro@example.com
```

ha un formato email valido.

Il browser controlla il formato, ma non verifica che l'indirizzo esista davvero.

### Il campo numerico

```html
<input
  type="number"
  id="eta"
  name="eta"
  min="0"
  max="120"
>
```

- `type="number"` indica un campo numerico.
- `min="0"` imposta il valore minimo valido.
- `max="120"` imposta il valore massimo valido.

Un valore fuori dall'intervallo rende il campo non valido e normalmente blocca l'invio del form.

Il campo è facoltativo perché non contiene `required`.

### L'attributo autocomplete

`autocomplete` aiuta il browser a proporre dati già conosciuti per compilare il form.

```html
autocomplete="given-name"
```

indica il nome della persona.

```html
autocomplete="email"
```

indica l'indirizzo email.

```html
autocomplete="street-address"
```

indica l'indirizzo stradale.

Il comportamento concreto dipende dal browser e dalle impostazioni dell'utente.

### Il pulsante di invio

```html
<button type="submit">Invia</button>
```

`type="submit"` indica che il pulsante serve a inviare il form.

Quando lo premi, il browser esegue prima i controlli HTML, come quelli attivati da `required`, `type="email"`, `min` e `max`.

## 7. Capire il CSS

### box-sizing: border-box

```css
* {
  box-sizing: border-box;
}
```

L'asterisco `*` seleziona tutti gli elementi.

Con `border-box`, padding e bordi sono compresi nella larghezza impostata.

Per esempio:

```css
input {
  width: 100%;
  padding: 10px;
  border: 1px solid #999;
}
```

L'input rimane dentro lo spazio disponibile nel form perché il suo `100%` comprende anche padding e bordi.

Senza `border-box`, con il valore predefinito `content-box`, padding e bordi si aggiungerebbero alla larghezza del contenuto e l'input potrebbe sporgere.

### Lo spazio intorno al form

```css
body {
  padding: 20px;
}
```

Aggiunge spazio interno intorno al contenuto della pagina.

Sugli schermi piccoli impedisce al form di toccare i bordi della finestra.

Dato che abbiamo applicato `box-sizing: border-box` anche al `body`, questo padding è compreso nella sua altezza minima di `100vh`.

### La larghezza del form

```css
form {
  width: 100%;
  max-width: 360px;
}
```

Le due proprietà lavorano insieme:

- `width: 100%` permette al form di usare lo spazio disponibile.
- `max-width: 360px` impedisce al form di superare 360 pixel.

Sul PC il form rimane compatto. Su uno schermo stretto può restringersi.

### Lo spazio interno del form

```css
form {
  padding: 24px;
}
```

Il `padding` crea spazio tra il bordo del form e il contenuto.

Non va confuso con il `margin`, che crea spazio all'esterno dell'elemento.

### Etichette sopra agli input

```css
label {
  display: block;
  margin-bottom: 6px;
  font-weight: bold;
}
```

- `display: block` fa occupare all'etichetta una riga, quindi l'input compare sotto.
- `margin-bottom: 6px` crea spazio tra etichetta e input.
- `font-weight: bold` rende il testo in grassetto.

### Input larghi quanto lo spazio disponibile

```css
input {
  width: 100%;
  padding: 10px;
}
```

`width: 100%` fa occupare all'input tutta la larghezza interna disponibile nel suo contenitore.

`padding: 10px` crea spazio tra il testo digitato e il bordo del campo.

### Ereditare il carattere

```css
input {
  font: inherit;
}

button {
  font: inherit;
}
```

`font: inherit` fa ereditare agli input e al pulsante le impostazioni del carattere dal loro elemento genitore.

In questo modo il testo del form mantiene uno stile coerente con quello della pagina.

### Evidenziare il campo selezionato

```css
input:focus {
  outline: 2px solid #155eab;
  outline-offset: 2px;
}
```

`:focus` si applica quando un input riceve il focus, per esempio quando ci clicchi o lo raggiungi con il tasto Tab.

- `outline` disegna un contorno blu.
- `outline-offset` distanzia il contorno dal bordo.

Questo aiuta a capire in quale campo stai scrivendo.

### Cambiare colore al passaggio del mouse

```css
button:hover {
  background-color: #104a87;
}
```

`:hover` si applica quando il puntatore si trova sopra il pulsante.

Il colore diventa più scuro per dare un riscontro visivo.

### Evidenziare il pulsante durante la navigazione da tastiera

```css
button:focus-visible {
  outline: 2px solid #155eab;
  outline-offset: 3px;
}
```

`:focus-visible` mostra il contorno quando il browser ritiene utile rendere evidente il focus, tipicamente durante la navigazione con la tastiera.

Puoi provarlo premendo Tab fino a raggiungere il pulsante.

### Il cursore sul pulsante

```css
button {
  cursor: pointer;
}
```

Mostra il cursore a forma di mano quando passi sul pulsante con il mouse.

## 8. Che cosa succede quando premi Invia?

Il form dell'esempio inizia così:

```html
<form method="get">
```

Non abbiamo specificato `action`, quindi il browser invia i dati all'indirizzo della stessa pagina.

Con `method="get"`, i valori vengono inseriti nell'URL come parametri.

Per esempio, l'indirizzo può terminare così:

```text
?nome=Alessandro&email=alessandro%40example.com&eta=55&indirizzo=Roma
```

`%40` rappresenta il simbolo `@` codificato nell'URL.

Questo esempio permette di capire la struttura e i controlli del form, ma **non salva i dati in un database**.

Per salvarli davvero servirà collegare il form a un backend, per esempio un server Express.

In un'applicazione reale, i dati devono essere verificati anche dal server: i controlli HTML nel browser aiutano l'utente, ma possono essere aggirati.
